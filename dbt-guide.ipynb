# Setting Up dbt-core

## Setting Up a Python venv

1. Install Python and setup the PATH
2. Open Up a directory for the Project
3. Run `python -m venv .venv` --will setup your new virtual environtment
4. Activate your venv by running `.venv/Scripts/activate` on your cmd
4. Install any python libraries and modules

## Install Dependencies

`pip install dbt-core` \
`pip install dbt-sqlserver` \
`pip install pyodbc` \
`pip install dbt-snowflake` -- if use snowflake

## Creating a new dbt project

Before you create a new dbt project make sure you have dbt installed: \
`dbt --version`

To create a new project run: \
`dbt init <project_name>` --skip-profile-setup (if you want to skip profile setup) \ 



## Setup your .env file and .gitignore

Checkout `.env.example` & `.gitignore.example`

To run dbt using the .env files without setting up on the `profiles.yml` do this: \
```
pip install python-dotenv[cli]

# use dotenv run in front of dbt command

dotenv run -- dbt debug
dotenv run -- dbt run
etc
```

# Create a script for dbt to read your .env file

```
cd <your_dbt_project>

# Create your own profiles.yml
<your_dbt_project/profiles.yml>

```

Checkout the example profiles.yml

## How dbt connect 

`dbt_project,yml` and `profiles.yml` 

The profile: value in dbt_project.yml: \
name: 'my_project'
profile: 'my_project'


```
dotenv run -- dbt run                   # uses dev (default) on the profiles.yml 
dotenv run -- dbt run --target prod     # uses prod
```
